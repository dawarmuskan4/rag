RAG - Retrieval Augmented Generation

it is the process of optimizing the output of a Large Language model, so it references an authoritative knowledge base outside of its training data sources before generating a response. LLMs are trained on vast volumne of data and use billions of parameters to generate original output for tasks like answering questions, translating languages and completing sentences. RAG extends the already powerful capabilities of LLms to specific domains or an organisation's internal knowledge base. All without the need to retrain the model. It is a cost-effective approach to improving LLM output so it remains relevant, accurate and useful in various context. 

First step is to parse the data - chunking

after you do chunking, you pass the data into embedding models. 
in embedding models, you basically convert text into vectors. vectors is like numerical representation for text so that you will be apply to apply algorithms like similarity search, cosine similarity techniques, that are already available - from where similar kind of information can be retrieved from the databases. This is called data ingestion pipeline in which you are able to store the text into vectors inside your vector DB. this is my knowledge base.


Data Ingestion Pipeline includes the step data ingetion, data parsing, embedding and all of this is stored in the vector store or vector DB. 
then the user query is also parsed into vectors which goes into vector db searched semantically adn the retrieval pipeline runs
Retrieval pipeline gets context from the semantic search in the vector db and that is given to the LLM which generates and output presented to the user.

augementation means you are giving context to the llm along with the prompt to generate the output.
